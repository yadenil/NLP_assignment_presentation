```python
# Question 1.2 — N-gram construction (n = 1, 2, 3, 4)
import json
from pathlib import Path

q1 = json.loads(Path("results/q1_results.json").read_text(encoding="utf-8"))
print("Question 1.2: N-gram Construction (n = 1, 2, 3, 4)")
for n in [1, 2, 3, 4]:
    key = f"top_{n}grams"
    print(f"\nTop {n}-grams:")
    for item in q1["top_ngrams"][key][:5]:
        print(item)
```

```python
# Question 1.3 — Top 10 most likely n-grams and probabilities
import json
from pathlib import Path

q1 = json.loads(Path("results/q1_results.json").read_text(encoding="utf-8"))
print("Question 1.3: Top 10 Most Likely N-grams")
for n in [1, 2, 3, 4]:
    key = f"top_{n}grams"
    print(f"\n--- {n}-grams ---")
    for item in q1["top_ngrams"][key][:10]:
        print(item)
```

```python
# Question 1.4 — Conditional probability of a word given context
import json
from pathlib import Path

q1 = json.loads(Path("results/q1_results.json").read_text(encoding="utf-8"))
print("Question 1.4: Conditional Probability Example")
for item in q1["top_ngrams"]["top_2grams"][:5]:
    print({
        "ngram": item["ngram"],
        "count": item["count"],
        "conditional_probability": item["conditional_probability"]
    })
```

```python
# Question 1.5 — Stopword removal and recomputed frequencies
import json
from pathlib import Path

q1 = json.loads(Path("results/q1_results.json").read_text(encoding="utf-8"))
print("Question 1.5: Stopword Removal and Recomputed Frequencies")
for n in [1, 2, 3, 4]:
    key = f"filtered_top_{n}grams"
    print(f"\n--- Filtered {n}-grams ---")
    for item in q1["top_filtered_ngrams"][key][:5]:
        print(item)
```

```python
# Question 1.6 — Word cloud generation and visualization
from pathlib import Path
from IPython.display import Image, display

images = sorted(Path("wordclouds").glob("*.png"))
print("Question 1.6: Word Cloud Visualizations")
for image in images:
    print(f"\n{image.name}")
    display(Image(filename=str(image)))
```

```python
# Question 1.7 — Sentence probability calculation
import json
from pathlib import Path

q1 = json.loads(Path("results/q1_results.json").read_text(encoding="utf-8"))
print("Question 1.7: Sentence Probability")
for entry in q1["sentence_probabilities"]:
    print(f"\n{entry['description']}:")
    print(entry["sentence"])
    for model_name, metrics in entry["models"].items():
        print(f"  {model_name}: log2_prob={metrics['log2_probability_smooth']:.4f}, smooth_prob={metrics['probability_smooth']:.6e}")
```

```python
# Question 1.8 — Random sentence generation and linguistic analysis
import json
from pathlib import Path

q1 = json.loads(Path("results/q1_results.json").read_text(encoding="utf-8"))
print("Question 1.8: Generated Sentences and Analysis")
print("\nGenerated sentences by model:")
for model_name, sentences in q1["generated_sentences"].items():
    print(f"{model_name}: {sentences[0] if sentences else 'No sentence generated'}")
print("\nAnalysis:")
print(q1["analysis"])
```

```python
# Question 2 — Intrinsic evaluation using perplexity and cross-entropy
import json
from pathlib import Path

q2 = json.loads(Path("results/q2_results.json").read_text(encoding="utf-8"))
print("Question 2: Intrinsic Evaluation")
for row in q2["results_table"][:10]:
    print(row)
print("\nDiscussion:")
print(q2["discussion"])
```

```python
# Question 3 — Extrinsic evaluation for next-word prediction and fluency discrimination
import json
from pathlib import Path

q3 = json.loads(Path("results/q3_results.json").read_text(encoding="utf-8"))
print("Question 3: Extrinsic Evaluation")
print("\nTask 1: Next-word prediction")
for name, res in q3["task1_next_word_prediction"].items():
    print(f"{name}: Top-1={res['top1_acc']:.2f}%, Top-3={res['top3_acc']:.2f}%, Top-5={res['top5_acc']:.2f}%, MRR={res['mrr']:.4f}")

print("\nTask 2: Fluency discrimination")
for name, res in q3["task2_fluency_discrimination"].items():
    print(f"{name}: accuracy={res['accuracy']:.2f}%")

print("\nDiscussion:")
print(q3["discussion"])
```
